```python
from google.colab import drive
try:
    drive.mount('/content/drive', force_remount=True)
except:
    pass

%pip install ultralytics

import zipfile
import os
import glob
from ultralytics import YOLO

# 1. Google Drive se dataset zip dhoondho
zips = glob.glob('/content/drive/**/*.zip', recursive=True)
zip_path = next((z for z in zips if 'taco' in z.lower() or 'trashnet' in z.lower() or 'yolov11' in z.lower()), zips[0] if zips else None)

if not zip_path:
    raise FileNotFoundError("Dataset ki zip file Google Drive par nahi mili!")

print(f"📁 Dataset zip mil gayi: {zip_path}")

# 2. Fast training ke liye dataset ko local Colab me extract karo
extract_path = '/content/dataset'
os.makedirs(extract_path, exist_ok=True)

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)
    print("📦 Dataset extracted successfully!")

# 3. data.yaml ka path fetch karo
data_yaml_paths = glob.glob(f'{extract_path}/**/data.yaml', recursive=True)
data_yaml = data_yaml_paths[0]
print("⚙️ Path to data.yaml:", data_yaml)

# 4. YOLOv11 nano model load karo
model = YOLO("yolo11n.pt")

# 5. Training start karo (Weights direct Google Drive par save honge!)
results = model.train(
    data=data_yaml,
    epochs=25,
    imgsz=640,
    batch=16,
    project='/content/drive/MyDrive/FYP/yolo_training_runs',
    name='ecorevive_yolov11_model_final'
)

print("🎉 Training successfully finished and saved to Google Drive!")
```

```python
import zipfile
import os
import glob
from ultralytics import YOLO

# 1. Dataset zip file dobara dhoondho
zips = glob.glob('/content/drive/**/*.zip', recursive=True)
zip_path = next((z for z in zips if 'taco' in z.lower() or 'trashnet' in z.lower() or 'yolov11' in z.lower()), zips[0] if zips else None)

if not zip_path:
    raise FileNotFoundError("Dataset ki zip file Google Drive par nahi mili!")

# 2. Extract dataset
extract_path = '/content/dataset'
os.makedirs(extract_path, exist_ok=True)

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

# 3. Get data.yaml path
data_yaml_paths = glob.glob(f'{extract_path}/**/data.yaml', recursive=True)
data_yaml = data_yaml_paths[0]
print("⚙️ Path to data.yaml:", data_yaml)

# 4. Load YOLOv11 Small (yolo11s.pt - bina 'v' ke!)
model = YOLO("yolo11s.pt")

# 5. Start Training with 60 Epochs
results = model.train(
    data=data_yaml,
    epochs=60,
    imgsz=640,
    batch=16,
    patience=20,
    project='/content/drive/MyDrive/FYP/yolo_training_runs',
    name='ecorevive_yolov11_60epochs'
)

print("🎉 60 Epochs Training successfully finished and saved to Google Drive!")
```
